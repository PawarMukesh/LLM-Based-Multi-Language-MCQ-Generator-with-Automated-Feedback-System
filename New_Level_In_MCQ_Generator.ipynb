# **Filter Answer**

```python
import pandas as pd
import re

# Read the Excel file
df = pd.read_excel('Generate_MCQ_Latest.xlsx')

# Define a function to extract only the options (A), (B), (C), and (D) from the Answer column
def extract_option(answer):
    match = re.search(r'\((A|B|C|D)\)', str(answer))
    if match:
        return match.group(0)
    else:
  # Replace values in the 'Answer' column using the mapping dictionary
        # Define a dictionary to map A, B, C, D to 1, 2, 3, 4 respectively
        mapping = {'A': 1, 'B': 2, 'C': 3, 'D': 4}
        df['Answer'] = df['Answer'].map(mapping)

# Apply the function to the 'Answer' column and overwrite it with the extracted options
df['Answer'] = df['Answer'].apply(extract_option)

# Save the modified dataframe to a new Excel file
df.to_excel('filtered_answers.xlsx', index=False)
```

```python
# Write the Streamlit app code in a Python script named app.py
import streamlit as st
from main import student_feedback

# Define the Streamlit app
def main():
    st.title("Student Feedback")

    excel_file_path = st.text_input("Excel File Path")

    programming_list = ['Python', 'Java', 'C & C++']
    language = st.selectbox("Select Programming Language:", programming_list)

    if st.button("Generate Feedback"):
        mcqs = student_feedback(excel_file_path, language)
        st.write("Feedback:")
        st.markdown(mcqs)

# Run the app
if __name__ == "__main__":
    main()
```
